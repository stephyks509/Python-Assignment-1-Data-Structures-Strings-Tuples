**1. String Concatenation:**

Write a Python program that takes two strings i.e string 1 “Hello ”, string 2 get name as input from the user and concatenates them together. Display the concatenated string as the output.

```python
# concatenate "Hello" with the user's name
Str1="Hello "
Str2=input("Enter your name ")
#get name as input from the user.
Output1=Str1+Str2
"""
concatenates Str1 and Str2. The + operator joins two strings without space.
 so Str1 ends in a space
 """
print(Output1) # prints Hello Stephy
```

**2. String Slicing and Indexing:**

Write a Python program using the above concatenated string as input and perform the
following tasks:

```python
I="Hello Stephy, welcome to Python programming"
#I is the inputed concatenated string
print(I)

```

```python
# a. Print the first character of the string I
a= I[0]

"""
In python,the index number of first character is assigned as 0,
 2nd character as 1 and so on.
 """

print(a)  # prints the first character H
```

```python
# b. Print the last character of the string.

b=I[-1]  # In python, the index number of last character is assigned as -1

print(b)  # prints the last character g
```

```python
# c. Print the first 5 characters of the string.  [start:stop]

c=I[:5]  #[:5] gives the starting character to the 5th character

print(c)  #prints Hello
```

```python
# d. Print the last 11 characters of the string.

d=I[-11:]

"""[-11] shows the -11th value from the last and [-11:] gives
the last 11 characters
"""

print(d)  #prints programming
```

```python
# e. Print the string in reverse.  [start:end:stepsize]
e=I[::-1]
#[::-1] will reverse the string charactors from end to the beginning
print(e)  #prints reversed string
```

```python
# f. Use slicing and print the word “Python” from the existing string.

f=I[25:31]

# in the above string the starting index of 'python' is 25 and ending is 31

print(f) #prints 'Python'
```

**3. String Methods:**

Write a Python program that takes a string, strM = “Python beginner tutorial” and
perform the following tasks:

```python
strM="Python beginner tutorial"
```

```python
# a. Convert the sentence to uppercase.

strM_uppercase=strM.upper() #string.upper()

print(strM_uppercase)  #prints string in Capital letters
```

```python
# b. Convert the sentence to lowercase.

strM_lowercase=strM.lower()  #string.lower()

print(strM_lowercase)  #prints string in small letters
```

```python
# c. Use Capitalize and return the sentence to the original input form.

strM_capitalise=strM_lowercase.capitalize()

"""
capitalize() converts the first character of the string to uppercase
 and others to the lowercase
 """

print(strM_capitalise)
```

```python
# d. Count the total number of occurrences of character ‘t’ in the string.

strM_count=strM_lowercase.count("t")
# for counting a character the string should be in lowercase or uppercase

print(strM_count)
#
'''
prints the number of times 't' occurs in string
strM="Python beginner tutorial"
'''
```

```python
'''
e. Replace all occurrences of “Python” with “Machine Learning” in the
input string strM=“Python beginner tutorial”
'''

strM_replace=strM.replace("Python","Machine Learning")

# .replace() replaces the existing string with new one.
# the string strM remains unchanged

print(strM_replace)  # Prints Machine Learning beginner tutorial
```

**Tuples (Creation, Modification and Access) :**

```python
"""
Create 1st tuple with values -> (10, 20, 30),
 2nd tuple with values -> (40, 50, 60):
 """
t1=(10,20,30)
t2=(40,50,60)
print(t1)
print(t2)
```

```python
# a. Concatenate the two tuples and store it in “t_combine”
t_combine=t1+t2  # + concatenates t1 and t2
print(t_combine)  #prints (10, 20, 30, 40, 50, 60)
```

```python
# b. Repeat the elements of “t_combine” 3 times
Repeat=t_combine*3  # *3 repeats the tuple 3 times in order
print(Repeat)
"""
prints (10, 20, 30, 40, 50, 60, 10, 20, 30, 40, 50, 60, 10, 20, 30, 40, 50, 60)
"""
```

```python
#c. Access the 3rd element from “t_combine”
Third_element=t_combine[2]
# the index of third element is 2, as the index number starts from 0
# Eg: tuple[2] gives the 3rd element from the tuple
print(Third_element)
```

```python
# d. Access the first three elements from “t_combine”
First_three_elements=t_combine[:3]
# t_combine[:3] gives the elements with 0 index to 2
print(First_three_elements)  #prints (10, 20, 30)
```

```python
# e. Access the last three elements from “t_combine”
Last_three_elements=t_combine[-3:]
"""
t_combine[-3:] gives the elements with -3 index to -1
thatis the last 3 elements
"""
print(Last_three_elements)  #prints (40, 50, 60)

```
